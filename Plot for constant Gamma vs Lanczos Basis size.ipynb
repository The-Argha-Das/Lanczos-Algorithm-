```python
import numpy as np
import matplotlib.pyplot as plt

gamma_true = np.array([953.6, 1040, 962.9, 1018, 903.4, 1102, 915.1, 1018,
                       940.6, 961, 929, 834.5, 1128, 1141, 1035, 974.9,
                       862.5, 1179, 1170, 1215, 832.7])
gamma_irlm = np.array([741.2, 804.3, 698.1, 721.4, 689.1, 805.2, 583.5, 774.9,
                       593.9, 697.8, 693, 611.5, 762.2, 700.5, 620.3, 640.7,
                       621.6, 753.9, 815.5, 810.7, 548.1])
basis = np.array([334.6, 331.6, 335.0, 352.4, 335.6, 363.4, 316.0, 347.6,
                  339.4, 333.6, 338.4, 327.6, 362.6, 342.8, 335.6, 329.4,
                  316.0, 376.4, 348.0, 379.8, 306.6])

mean_basis = basis.mean()
std_basis = 2 * basis.std()

# Plot basis vs gamma_true with horizontal mean + 1σ band
plt.figure()
plt.scatter(gamma_true, basis)
plt.axhline(mean_basis, linestyle='-', label=f'mean basis ≈ {mean_basis:.1f}')
plt.axhline(mean_basis + std_basis, linestyle='--', label='±2σ band')
plt.axhline(mean_basis - std_basis, linestyle='--')
plt.xlabel('γ(true)')
plt.ylabel('Lanczos basis size')
plt.title('Constant-γ experiment: basis vs γ(true)\n(basis stays within ~±5% of mean)')
plt.legend()
plt.grid(True)
plt.show()

# Plot basis vs γ(IRLM) similarly
plt.figure()
plt.scatter(gamma_irlm, basis)
plt.axhline(mean_basis, linestyle='-', label=f'mean basis ≈ {mean_basis:.1f}')
plt.axhline(mean_basis + std_basis, linestyle='--', label='±2σ band')
plt.axhline(mean_basis - std_basis, linestyle='--')
plt.xlabel('γ(IRLM)')
plt.ylabel('Lanczos basis size')
plt.title('Constant-γ experiment: basis vs γ(IRLM)\n(basis stays within ~±5% of mean)')
plt.legend()
plt.grid(True)
plt.show()
```
