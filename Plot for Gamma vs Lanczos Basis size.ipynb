```python
import matplotlib.pyplot as plt
import numpy as np

gamma_true = np.array([1630.16, 1174.62, 1547.24, 1562.12, 1132.02, 1229.38, 1277.71, 1351.61,
                       1712.06, 1670.77, 1255.93, 1131.39, 1319.78, 1341.52, 1050.25, 1339.41,
                       1099.25, 1515.02, 1029.58, 2353.16, 1324.5])
gamma_irlm = np.array([832.778, 737.818, 785.079, 972.375, 754.918, 866.074, 849.753, 783.602,
                       900.125, 920.025, 727.802, 693.212, 788.82, 774.863, 637.131, 767.865,
                       738.883, 858.498, 647.106, 1554.65, 765.803])
basis = np.array([402.35, 362.0, 402.35, 385.275, 347.05, 360.55, 367.25, 381.2,
                  416.15, 374.425, 365.5, 351.65, 377.05, 363.95, 343.85, 372.9,
                  351.7, 393.75, 341.3, 405.475, 380.2])

coef_true = np.array([11.60428914, -2952.32868129])
coef_irlm = np.array([5.06522705, -1065.91483629])

x_b = np.linspace(basis.min(), basis.max(), 100)
y_true_fit = np.polyval(coef_true, x_b)
y_irlm_fit = np.polyval(coef_irlm, x_b)

# Plot 1: γ(true) vs basis
plt.figure()
plt.scatter(basis, gamma_true)
plt.plot(x_b, y_true_fit)
plt.xlabel('Average Lanczos basis size')
plt.ylabel('Average γ(true)')
plt.title('Linear relation: γ(true) ≈ 11.6·basis − 2952')
plt.grid(True)
plt.show()

# Plot 2: γ(IRLM) vs basis
plt.figure()
plt.scatter(basis, gamma_irlm)
plt.plot(x_b, y_irlm_fit)
plt.xlabel('Average Lanczos basis size')
plt.ylabel('Average γ(IRLM)')
plt.title('Linear relation: γ(IRLM) ≈ 5.07·basis − 1066')
plt.grid(True)
plt.show()
```
